# R07-H37 - is PPR theater at this scale? Ranked expansion vs a plain 1-hop MATCH

**Author**: Knowledge Graph Foundry autonomous build (kj) <br>
**Date**: 2026-07-06 <br>
**Pipeline stage**: R07 contrarian slate, completion-free set (H34 pre-signal: PPR channel carried 0 golds) <br>
**Graph**: rebuilt CPAP corpus (neo4j2), Titan embeddings, zero completions <br>

PPR's value proposition is global structure discovery, adopted (R01-H2) from literature on graphs 100-1000x larger. On a <5k-entity graph with seed-local questions, the contrarian claim is that PPR's ranked expansion is a 1-hop neighborhood in an expensive costume - the damping, GDS projection and PageRank iteration buy nothing a MATCH clause doesn't.

## Approach
1. **Harvest** - per probe: production seed set (vector top-k + proposition seeds), then (a) PPR top-n expansion and (b) the plain 1-hop neighborhood of the seeds (non-SIMILAR_TO, currently-valid edges)
2. **Containment** - fraction of PPR-returned nodes that are seeds or 1-hop of seeds (pre-registered: >= 80%)
3. **Evidence swap** - gold strings found in the render of PPR-beyond-1-hop nodes vs 1-hop-only nodes; the refuter is PPR-exclusive gold on >= 2 probes
4. **Verdict** - against the pre-registered bar

## Outputs
- `reports/probe-eval-r07h37-<stamp>.json` - per-probe containment, expansion sizes, gold placement, verdict
- In-notebook: containment distribution, gold-placement table, verdict

```python
# Imports
# stdlib
import datetime  # report stamps
import json  # report persistence
import os  # graph selection env
import re  # gold matching

# third party
import yaml  # probe set
from pathlib import Path
from rich import print as rprint  # semantic output
from rich.progress import Progress  # harvest loop

os.environ["NEO4J_URI"] = "bolt://user-konrad.jelen-kgf-neo4j2:7687"
os.environ["NEO4J_USER"] = "neo4j"
os.environ["NEO4J_PASSWORD"] = "kgfoundry"

# project
from knowledge_graph_foundry import Foundry, load_settings  # pipeline
from knowledge_graph_foundry.extraction import generate_embeddings  # Titan probe embedding
from knowledge_graph_foundry.graph.graphrag import ppr_query, vector_query  # retrieval channels
from knowledge_graph_foundry.graph.propositions import proposition_query  # proposition seeds
from knowledge_graph_foundry.models import Entity  # embedding probe wrapper
```

```python
# Configuration
PROBES_PATH = Path("../tests/probes/cpap-probe-set.yml")  # 28 probes with gold evidence

settings = load_settings(Path("../config.yml"))
settings.graphrag.propositions_enabled = True  # production channel state

TOP_K = settings.graphrag.top_k                        # vector seeds per probe
PPR_TOP_N = settings.graphrag.ppr_top_n                # PPR nodes into context
PPR_DAMPING = settings.graphrag.ppr_damping            # PageRank damping
PROP_TOP_K = settings.graphrag.proposition_top_k       # propositions per probe
VEC_INDEX = settings.graphrag.vector_index_name        # entity vector index
PROP_INDEX = settings.graphrag.proposition_index_name  # proposition vector index

# pre-registered acceptance bar (experiments log, R07-H37)
BAR_CONTAINMENT = 0.80   # PPR nodes inside seeds+1hop
BAR_REFUTE_PROBES = 2    # PPR-exclusive gold on this many probes refutes

probes = yaml.safe_load(PROBES_PATH.read_text())
gold_probes = [p for p in probes if p.get("gold_evidence")]

rprint(f"""[bold cyan]Configuration[/bold cyan]
[dim]{"\u2500" * 40}[/dim]
[bold]Graph[/bold]
  Neo4j: [cyan]{os.environ['NEO4J_URI']}[/cyan]

[bold]Channels (production parameters)[/bold]
  Vector top-k: [yellow]{TOP_K}[/yellow]  PPR top-n: [yellow]{PPR_TOP_N}[/yellow] [dim](damping {PPR_DAMPING})[/dim]  Proposition top-k: [yellow]{PROP_TOP_K}[/yellow]

[bold]Probes[/bold]
  Set: [cyan]{PROBES_PATH}[/cyan] ([yellow]{len(gold_probes)}[/yellow] with gold evidence)

[bold]Acceptance bar[/bold]
  Confirmed: containment >= [yellow]{BAR_CONTAINMENT}[/yellow] and no probe-outcome change under swap
  Refuted: PPR-exclusive gold on >= [yellow]{BAR_REFUTE_PROBES}[/yellow] probes
""")


def _norm(s):
    return re.sub(r"\s+", " ", s.casefold())


# gold matcher - the H34 iteration-3 conventions (substring, unit-stripped
# numeric skeleton, value-token majority, content-word overlap)
_UNIT = r"(?<=\d)\s*(mm|cm|dba|db\(a\)|db|kg|g|oz|ml|l|w|hz|mins|min|m)\b"


def present(gold, ctx_norm):
    ng = _norm(gold)
    if ng in ctx_norm:
        return True
    squashed = re.sub(r"[\s,()]", "", ctx_norm)
    skeleton = re.sub(r"[\s,()]", "", re.sub(_UNIT, "", ng))
    if any(ch.isdigit() for ch in skeleton) and len(skeleton) >= 5 and skeleton in squashed:
        return True
    tokens = re.findall(r"[\w.\-/]*\d[\w.\-/]*", gold)
    if tokens:
        hit = sum(1 for t in tokens if _norm(t) in ctx_norm or re.sub(r"[\s,()]", "", _norm(t)) in squashed)
        return hit >= max(1, len(tokens) // 2 + (len(tokens) % 2))
    words = set(re.findall(r"[a-z][a-z0-9\-]{2,}", ng))
    ctx_words = set(re.findall(r"[a-z][a-z0-9\-]{2,}", ctx_norm))
    return bool(words) and len(words & ctx_words) / len(words) >= 0.6
```

## Harvest, containment and gold placement

Per probe: production seeds (vector + proposition), PPR expansion, 1-hop neighborhood of the seeds. Containment = |PPR result inside seeds+1hop| / |PPR result|. Gold placement renders three disjoint node sets with the production render (alias merge on) and tests each gold string: PPR-beyond-1-hop nodes (the only nodes PPR uniquely contributes), 1-hop-beyond-PPR nodes (what a MATCH adds that PPR ranked out), and their intersection.

```python
def render_nodes(session, node_ids):
    """Production per-node render (description + prop_* spec + alias merge +
    currently-valid relations) - same as probe_eval_r07h34."""
    blocks = []
    for nid in node_ids:
        row = session.run(
            "MATCH (e:Entity {id: $id}) RETURN e.name AS name, labels(e) AS types, "
            "e.description AS description, properties(e) AS props",
            id=nid,
        ).single()
        if row is None:
            continue
        spec = {k.removeprefix("prop_"): v for k, v in row["props"].items() if k.startswith("prop_")}
        aliases = session.run(
            "MATCH (e:Entity {id: $id})-[:SAME_AS*1..2]-(a:Entity) "
            "WHERE a.id <> $id RETURN DISTINCT a.name AS name, properties(a) AS props LIMIT 5",
            id=nid,
        ).data()
        for a in aliases:
            for k, v in a["props"].items():
                if k.startswith("prop_"):
                    spec.setdefault(k.removeprefix("prop_"), v)
        rels = session.run(
            "MATCH (e:Entity {id: $id})-[r]-(n:Entity) "
            "WHERE r.valid_to IS NULL AND type(r) <> 'SIMILAR_TO' "
            "RETURN type(r) AS rel, n.name AS name LIMIT 15",
            id=nid,
        ).data()
        blocks.append(
            f"## {row['name']} ({', '.join(row['types'])})\n"
            + (f"Also known as: {', '.join(a['name'] for a in aliases)}\n" if aliases else "")
            + f"{row['description'] or ''}\n"
            f"Properties: {json.dumps(spec, default=str)}\n"
            "Relations: " + "; ".join(f"{r['rel']} -> {r['name']}" for r in rels)
        )
    return "\n".join(blocks)


results = {}
with Foundry(settings) as f, Progress() as progress:
    task = progress.add_task("PPR vs 1-hop", total=len(gold_probes))
    for p in gold_probes:
        q = p["question"]
        probe_e = Entity.create(q[:80], types=["Query"], description=q)
        emb = generate_embeddings([probe_e], settings.embeddings)[0].embedding
        seeds = vector_query(f.driver, emb, VEC_INDEX, top_k=TOP_K)
        seed_ids = [s["id"] for s in seeds]
        prop_hits = proposition_query(f.driver, emb, PROP_INDEX, top_k=PROP_TOP_K)
        for h in prop_hits:
            for eid in h["entity_ids"]:
                if eid not in seed_ids:
                    seed_ids.append(eid)

        ranked = ppr_query(f.driver, seed_ids, top_n=PPR_TOP_N, damping=PPR_DAMPING)
        ppr_ids = [n["id"] for n in ranked]
        with f.driver.session() as session:
            onehop = {
                r["id"] for r in session.run(
                    "MATCH (e:Entity)-[r]-(n:Entity) WHERE e.id IN $ids "
                    "AND r.valid_to IS NULL AND type(r) <> 'SIMILAR_TO' "
                    "RETURN DISTINCT n.id AS id",
                    ids=seed_ids,
                )
            }
            seedset = set(seed_ids)
            inside = [i for i in ppr_ids if i in seedset or i in onehop]
            ppr_beyond = [i for i in ppr_ids if i not in seedset and i not in onehop]
            onehop_beyond = list(onehop - set(ppr_ids) - seedset)
            ctx_ppr_beyond = _norm(render_nodes(session, ppr_beyond))
            ctx_onehop_beyond = _norm(render_nodes(session, onehop_beyond))

        golds_ppr = [g for g in p["gold_evidence"] if present(g, ctx_ppr_beyond)]
        golds_onehop = [g for g in p["gold_evidence"] if present(g, ctx_onehop_beyond)]
        results[p["id"]] = {
            "containment": len(inside) / len(ppr_ids) if ppr_ids else 1.0,
            "n_ppr": len(ppr_ids), "n_ppr_beyond": len(ppr_beyond),
            "n_onehop": len(onehop), "n_onehop_beyond": len(onehop_beyond),
            "gold_in_ppr_beyond": golds_ppr, "gold_in_onehop_beyond": golds_onehop,
        }
        progress.advance(task)

rprint(f"[green]harvested[/green] {len(results)} probes")
```

## Verdict

Mean containment against the 80% bar; refuter count = probes where a gold string lives ONLY in PPR-beyond-1-hop nodes. The complementary count (gold in 1-hop-beyond-PPR) shows what PPR's ranking actively discards.

```python
mean_containment = sum(r["containment"] for r in results.values()) / len(results)
ppr_exclusive_probes = [pid for pid, r in results.items() if r["gold_in_ppr_beyond"]]
onehop_exclusive_probes = [pid for pid, r in results.items() if r["gold_in_onehop_beyond"]]

refuted = len(ppr_exclusive_probes) >= BAR_REFUTE_PROBES
confirmed = mean_containment >= BAR_CONTAINMENT and not refuted
verdict = "REFUTED" if refuted else ("CONFIRMED" if confirmed else "INCONCLUSIVE")

lows = sorted(results.items(), key=lambda kv: kv[1]["containment"])[:5]
rprint(f"""[bold cyan]PPR vs 1-hop[/bold cyan]
[dim]{"\u2500" * 40}[/dim]
  Mean containment (PPR inside seeds+1hop): [yellow]{mean_containment:.3f}[/yellow] [dim](bar >= {BAR_CONTAINMENT})[/dim]
  Lowest-containment probes: {[(pid, round(r['containment'], 2)) for pid, r in lows]}
  Probes with gold ONLY reachable by PPR-beyond-1-hop: [yellow]{ppr_exclusive_probes}[/yellow] [dim](refute at >= {BAR_REFUTE_PROBES})[/dim]
  Probes with gold in 1-hop nodes PPR ranked OUT: [yellow]{onehop_exclusive_probes}[/yellow]
  Verdict: [{'green' if verdict == 'CONFIRMED' else 'red' if verdict == 'REFUTED' else 'yellow'}]{verdict}[/]
""")

stamp = datetime.datetime.now(datetime.timezone.utc).strftime("%Y%m%d-%H%M%S")
out = Path("../reports") / f"probe-eval-r07h37-{stamp}.json"
out.write_text(json.dumps({
    "mean_containment": mean_containment,
    "ppr_exclusive_probes": ppr_exclusive_probes,
    "onehop_exclusive_probes": onehop_exclusive_probes,
    "per_probe": results, "verdict": verdict,
}, indent=2, default=str))
rprint("saved", str(out))
```
